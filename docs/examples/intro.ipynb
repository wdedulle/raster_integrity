```python
print("Hello World!")
from osgeo import gdal

gdal.UseExceptions()

gdal.Translate(
    r"..\..\data\classified.png",
    r"..\..\data\classified.jpg",
    format="PNG",
    bandList=[1],
)
gdal.Translate(
    r"..\..\data\classified.tif",
    r"..\..\data\classified.jpg",
    format="GTiff",
    bandList=[1],
)
gdal.Translate(
    r"..\..\data\classified.asc",
    r"..\..\data\classified.jpg",
    format="AAIGrid",
    bandList=[1],
)
```
